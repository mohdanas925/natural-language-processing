# 3. N-Grams

```python
from nltk.util import ngrams
from nltk.tokenize import word_tokenize

text = "machine learning is interesting"

words = text.split()

# Unigrams
unigrams = list(ngrams(words, 1))
print("Unigrams:", unigrams)

# Bigrams
bigrams = list(ngrams(words, 2))
print("Bigrams:", bigrams)

# Trigrams
trigrams = list(ngrams(words, 3))
print("Trigrams:", trigrams)
```

```python
bigrams = [
    " ".join(gram)
    for gram in ngrams(words, 2)
]

print(bigrams)
```

### Appying it on dataframe

```python
import pandas as pd

df1 = pd.DataFrame({
    "description_text": [
        "Required skills include Python, SQL, and machine learning.",
        "Experience in data analysis, data visualization, and Excel.",
        "Develop NLP models using Python and natural language processing.",
        "Knowledge of software development, debugging, and testing.",
        "Work with deep learning, neural networks, and computer vision."
    ]
})

df1
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Remove missing values and convert text to strings
text_data = df1["description_text"].fillna("").astype(str)

vectorizer = CountVectorizer(
    ngram_range=(1, 3),
    stop_words="english",
    max_features=5000
)

X = vectorizer.fit_transform(text_data)

# View the extracted features
ngrams_list = vectorizer.get_feature_names_out()

print(ngrams_list[:50])
```

```python
# Finding most-frequent n-grams across dataset
import pandas as pd

ngram_counts = X.sum(axis=0).A1

ngram_frequency = pd.DataFrame({
    "ngram": vectorizer.get_feature_names_out(),
    "frequency": ngram_counts
})

ngram_frequency = ngram_frequency.sort_values(
    by="frequency",
    ascending=False
)

print(ngram_frequency.head(5))
```

# 4. TF-IDF 

```python
import pandas as pd

df = pd.DataFrame({
    "description_text": [
        "python machine learning",
        "python data analysis",
        "machine learning data science"
    ]
})

df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TF-IDF
tfidf = TfidfVectorizer()

# Convert text into numerical features
X = tfidf.fit_transform(df["description_text"])

# Convert to DataFrame for easy viewing
tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=tfidf.get_feature_names_out()
)

print(tfidf_df.round(3))
```

```python
# Finding most-important words from each document
feature_names = tfidf.get_feature_names_out()

for i, row in enumerate(X.toarray()):
    top_indices = row.argsort()[::-1][:3]

    top_words = [
        (feature_names[j], round(row[j], 3))
        for j in top_indices
        if row[j] > 0
    ]

    print(f"Document {i}: {top_words}")
```
