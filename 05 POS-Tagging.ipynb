### Applying POS on Textual Data

```python
import nltk

nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")

from nltk.tokenize import word_tokenize
from nltk import pos_tag

text = """
We are looking for a skilled Python developer
with experience in machine learning and data analysis.
"""

words = word_tokenize(text)
tagged_words = pos_tag(words)

print(tagged_words)
```

### Applying POS on Dataframe

```python
import pandas as pd
df = pd.DataFrame({
    "description_text": [
        "We need a Python developer.",
        "The candidate should know machine learning.",
        "Build and maintain data pipelines."
    ]
})

df["pos_tags"] = df["description_text"].apply(
    lambda text: pos_tag(word_tokenize(text))
)

print(df[["description_text", "pos_tags"]])
```

```python
def extract_pos(text):
    words = word_tokenize(text)
    tagged = pos_tag(words)

    nouns = [word for word, tag in tagged
             if tag.startswith("NN")]

    verbs = [word for word, tag in tagged
             if tag.startswith("VB")]

    adjectives = [word for word, tag in tagged
                  if tag.startswith("JJ")]

    return pd.Series({
        "nouns": nouns,
        "verbs": verbs,
        "adjectives": adjectives
    })

df[["nouns", "verbs", "adjectives"]] = (
    df["description_text"].apply(extract_pos)
)

df
```

### Applying POS on List

```python
story = [
    ["the", "wildlings", "are", "dead"],
    ["do", "the", "dead", "frighten", "you"]
]
```

```python
from nltk import pos_tag

tagged_story = [pos_tag(sentence) for sentence in story]

print(tagged_story)
```

## Hidden Markov Model POS Tagging

```python
import nltk
from nltk.corpus import brown
from nltk.tag.hmm import HiddenMarkovModelTrainer
```

```python
nltk.download("brown")

tagged_sentences = brown.tagged_sents()

print(tagged_sentences[0])
```

```python
trainer = HiddenMarkovModelTrainer()

hmm_model = trainer.train_supervised(
    tagged_sentences[:40000]
)
```

```python
sentence = "The student learns Python"

words = nltk.word_tokenize(sentence)

tagged_words = hmm_model.tag(words)

print(tagged_words)
```
