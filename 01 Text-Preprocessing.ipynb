## 2. Remove HTML Tags

```python
from bs4 import BeautifulSoup

text = "<h1>Welcome to NLP</h1><p>This is a <b>great</b> topic.</p>"

clean_text = BeautifulSoup(text, "html.parser").get_text()

print(clean_text)
```

```python
df = pd.DataFrame({
    "text": [
        "<h1>Hello</h1>",
        "<p>I love <b>NLP</b></p>",
        "<div>Python is <i>amazing</i></div>"
    ]
})

df
```

```python
import re

def remove_html_tags(text):
    # Remove HTML tags using Regex
    text = re.sub(r'<[^>]+>', '', text)

    # Remove remaining HTML tags using BeautifulSoup
    text = BeautifulSoup(text, "html.parser").get_text()

    return text


# Apply function to DataFrame column
df["text"] = df["text"].apply(remove_html_tags)

df 
```

## 3. Remove URLs

```python
df = pd.DataFrame({
    "text": [
        "Visit https://www.google.com for more information.",
        "Check out https://github.com/mohdanas925",
        "Visit www.example.com to learn NLP.",
        "Follow us at https://www.linkedin.com/in/example",
        "This text does not contain any URL."
    ]
})

df
```

```python
import re

# Function to remove URLs
def remove_urls(text):
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    return text


# Apply function to DataFrame column
df["text"] = df["text"].apply(remove_urls)

df
```

```python
def chat_conversion(text):
    new_text = []
    for word in text.split():
        if word.upper() in chat_words:
            new_text.append(chat_words[word.upper()])
        else:
            new_text.append(word)
    return " ".join(new_text)
```

```python
emoji_dict = {
    "😍": "love",
    "🔥": "excellent",
    "😢": "sad",
    "😂": "laughing",
    "😊": "happy",
    "😡": "angry",
    "❤️": "love",
    "👍": "good"
}

text = "I love this movie 😍🔥 but the ending was sad 😢"

for emoji_char, meaning in emoji_dict.items():
    text = text.replace(emoji_char, "" + meaning + " ")

print(text)
```

## 10. Stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "plays", "studies", "studying", "running"]

for word in words:
    print(word, "→", stemmer.stem(word))
```

```python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

stemmer = PorterStemmer()

text = "I am playing games and studying machine learning"

words = word_tokenize(text)

stemmed_words = [stemmer.stem(word) for word in words]

print(stemmed_words)
```

```python
# Stemmin on DataFrame
import pandas as pd
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

stemmer = PorterStemmer()

df = pd.DataFrame({
    "text": [
        "I am playing football",
        "He played football yesterday",
        "She is studying machine learning"
    ]
})

def stem_text(text):
    words = word_tokenize(text)
    return " ".join(stemmer.stem(word) for word in words)

df["stemmed_text"] = df["text"].apply(stem_text)

print(df)
```

## 11. Lemmatization

```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")

lemmatizer = WordNetLemmatizer()

words = ["studies", "playing", "cars", "children", "running"]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word))
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("studies", pos="n"))
print(lemmatizer.lemmatize("studying", pos="v"))
print(lemmatizer.lemmatize("running", pos="v"))
print(lemmatizer.lemmatize("better", pos="a"))
print(lemmatizer.lemmatize("children", pos="n"))
```

```python
# Using SpaCy

import spacy

nlp = spacy.load("en_core_web_sm")

text = "The children are running and studying different subjects."

doc = nlp(text)

for token in doc:
    print(token.text, "→", token.lemma_)
```

```python
# Lemmatization on DataFrame
import pandas as pd
import spacy

nlp = spacy.load("en_core_web_sm")

df = pd.DataFrame({
    "text": [
        "The children are playing games",
        "He is studying machine learning",
        "The cars are running quickly"
    ]
})

def lemmatize_text(text):
    doc = nlp(text)
    return " ".join(token.lemma_ for token in doc)

df["lemmatized_text"] = df["text"].apply(lemmatize_text)

print(df)
```
